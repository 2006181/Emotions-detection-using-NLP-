```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train.txt', sep=';', header=None, names=['text', 'emotion'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions = df['emotion'].unique()
emotion_number = {}
i = 0
for emotion in unique_emotions:
    emotion_number[emotion] = i
    i += 1

df['emotion'] = df['emotion'].map(emotion_number)    
```

```python
df
```

```python
df['text'] = df['text'].str.lower()
df
```

```python
import string

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))    
```

```python
df.text = df['text'].apply(remove_punctuation)  
```

```python
def remove_numbers(text):
    new = ""
    for i in text:
        if not i.isdigit():
            new = new + i
    return new  

df.text = df['text'].apply(remove_numbers)  
```

```python
import emoji

def remove_emojis(text):
    new = ""

    for i in text:
        if not emoji.is_emoji(i):
            new = new + i

    return new

df.text = df['text'].apply(remove_emojis)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize      
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(text):
    words = text.split()
    cleaned = []
    for i in words:
        if i not in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)    
```

```python
df['text'] = df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split    

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer 

bow_vectorizer = CountVectorizer()

X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

nb = MultinomialNB()
nb = nb.fit(X_train_bow, y_train)   
predictions = nb.predict(X_test_bow)

print("Accuracy:", accuracy_score(y_test, predictions))
nb.fit(X_train_bow, y_train)

```

```python
def predict_emotion(text):

    text = text.lower()
    text = remove_punctuation(text)
    text = remove_numbers(text)
    text = remove_emojis(text)
    text = remove(text)

    text_bow = bow_vectorizer.transform([text])

    prediction = nb.predict(text_bow)

    emotion = list(emotion_number.keys())[prediction[0]]

    return emotion
```

```python
predict_emotion("I am feeling very happy today")
```

```python
import pickle

# Save trained model
with open("emotion_model.pkl", "wb") as file:
    pickle.dump(nb, file)

# Save BOW vectorizer
with open("bow_vectorizer.pkl", "wb") as file:
    pickle.dump(bow_vectorizer, file)

with open("emotion_number.pkl", "wb") as file:
    pickle.dump(emotion_number, file)    

print("Model and vectorizer saved successfully!")
```
